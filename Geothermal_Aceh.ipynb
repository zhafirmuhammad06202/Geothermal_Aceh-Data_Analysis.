```python
import pandas as pd
import numpy as np

data_geothermal = {
    'Lokasi': ['Seulawah Agam', 'Jaboi Sabang', 'Ie Seu-um', 'Pulo Aceh', 'Gayo Lues', 'Bener Meriah'],
    'Lat': [5.448, 5.815, 5.545, 5.612, 3.965, 4.725],
    'Lon': [95.658, 95.332, 95.542, 95.125, 97.352, 96.852],
    'Temp_C': [98, 105, 85, 45, 92, 88],
    'pH': [2.5, 3.2, 7.5, 8.1, 6.8, 5.5],
    'Silica_ppm': [150, 210, 80, 25, 120, 110],
    'Tipe': ['Vulkanik', 'Vulkanik', 'Sedimentari', 'Non-Volcanic', 'Vulkanik', 'Vulkanik']
}

df_geo = pd.DataFrame(data_geothermal)
print ("data berhasil dibuat")
print (df_geo.head())

import plotly.express as px
fig = px.scatter(df_geo, x="Silica_ppm", y="Temp_C",
                 size = "Temp_C", color = "Tipe",
                 hover_name="Lokasi",
                 title ="hubungan silica terhadap temperatur panas di tanah aceh",
                 labels ={"Silica_ppm": "Kadar Silica (ppm)", "Temp_C": "suhu celcius (C)"})

fig.show()

import folium

peta_aceh = folium.Map(location=[5.0, 96.5], zoom_start=8, tiles='OpenStreetMap')

for index, row in df_geo.iterrows():

    warna = 'red' if row['Temp_C'] > 90 else 'orange' if row['Temp_C'] > 60 else 'blue'

    folium.Marker(
        location=[row['Lat'], row['Lon']],
        popup=f"Lokasi: {row['Lokasi']}\nSuhu: {row['Temp_C']}°C\npH: {row['pH']}",
        tooltip=row['Lokasi'],
        icon=folium.Icon(color=warna, icon='info-sign')
    ).add_to(peta_aceh)

peta_satelit = folium.Map(location=[5.2, 96.5], zoom_start=8, tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}', attr='Esri')

for index, row in df_geo.iterrows():
    warna = 'red' if row['Temp_C'] > 90 else 'orange' if row['Temp_C'] > 60 else 'blue'
    folium.Marker(
        location=[row['Lat'], row['Lon']],
        popup=f"<b>{row['Lokasi']}</b><br>Suhu: {row['Temp_C']}°C<br>pH: {row['pH']}",
        icon=folium.Icon(color=warna)
    ).add_to(peta_satelit)

peta_satelit.save('peta_satelit_geothermal.html')
print("Peta Satelit Berhasil Dibuat!")
```
